```python
def iter_thrgh_dict(report_as_list,values_as_dict): #на вход list

    for i_dict in report_as_list: #итерируем по словарям в списке
        if 'value' in  i_dict:
            i_dict['value']=values_as_dict[i_dict['id']]
            print(i_dict['id'],":",values_as_dict[i_dict['id']])
        if 'values' in i_dict:
                print("..заходим во вложенность типа ",type(i_dict['values']))
                try: # опять итерируем по словарям в новом списке
                    iter_thrgh_dict(i_dict['values'],values_as_dict)
                except KeyError as ww:
                    print(f"опять проблема с {ww}") #итерируем по словарям в списке
        else:print('..всё, вышли из вложенности')
    return report_as_list
```
